## Encapsulation
biding of data into single unit(attributes,methods)

inorder to hide the data to handle the methods

we are using getter and setter method in order access the private varaible

```python
class Bankaccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.__balance=balance
    ## getter method to access balance
    def get_balance(self):
        return self.__balance
    ## setter method to modify balance(only allow through depoit And withdraw)
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print(f"deposited {amount}.new balance: {self.__balance}")
        else:
            print("Invalid depoit amount.")
    def withdraw(self,amount):
        if 0<amount<=self.__balance:
            self.__balance-=amount
            print(f"withdraw {amount}.new balance:{self.__balance}")
        else:
            print("invalid withdrawal amount or insufficient balance.")
            
        
```

##### getter - access the balance
##### setter -modify the balance

```python
b2=Bankaccount("radhika",5000)
```

```python
b2.get_balance()
```

```python
b2.deposit(2000)
```

```python
b2.withdraw(200)
```

```python
b2.withdraw(30)
```

```python
b2.withdraw(6770)
```

```python
b2.withdraw(50)
```

```python
class car:
    def __init__(self,model):
        self.model=model
        self.__speed=0
        
    def get_speed(self):
        return self.__speed
        
    def set_speed(self,speed):
        if 0<=speed<=200:
            self.__speed=speed
            print(f"speed set to {speed} km/hr")
        else:
            print("invalid speed! Enter a value between 0 and 200 km/h.")
            
    def acceleration(self,increment):
        new_speed=self.__speed+increment
        self.set_speed(new_speed)
        
    def decrement(self,decrement):
        new_speed=self.__speed-decrement
        self.set_speed(new_speed)
        
```

```python
obj1=car("bmw")
```

```python
obj1.set_speed(100)
```

```python
obj1.get_speed()
```

```python
obj1.acceleration(100)
```

```python
obj1.acceleration(29)
```

```python
obj1.decrement(50)
```

```python
from abc import ABC,abstractmethod ##ABC means abstractionn based class
```

```python
class animal(ABC): #in parent class the method is already implemented this is the diffenece betwwen abstractuion and polummorphism
        @abstractmethod
        def make_sound(self):
            pass # even though we are not implementing we can access in the parent class

class Dog(animal):
    def make_sound(self):
        return "the dog barks"

class cat(animal):
    def make_sound(self):
        return "cat meows"
```

```python
class shape(ABC):
    @abstractmethod
    def area(self):
        pass
        
class square(shape):
    def __init__(self,side):
        self.side=side
        
    def area(self):
        return self.side ** 2
        
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
        
    def area(self):
        return 3.14 * self.radius **2 
```

```python
o=square(4)
```

```python
o.area()
```

```python
b=circle(2)
```

```python
b.area()
```
